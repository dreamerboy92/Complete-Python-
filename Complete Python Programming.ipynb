```python
def grade_distribution(grades):
    for grade in grades:
        if 90<= grade <=100:
            print("You have A grade")
        elif 80<= grade <=89:
            print("You have got B grade")
        elif 70<= grade <=79:
            print("You have got C grade")
        elif 60<= grade <=69:
            print("You have got D grade")
        else:
            print("You have got E grade")
            return 
```
